### string methods
- Case conversion methods
    - upper()
    - lower()
    - title()
    - swapcase()
    - capitalize()
  

```python
a="jewlch jefwqhfuibc 2"
b=a.upper()
c=a.lower()
print(b)
print(a.capitalize())
print(c)


```

```python
##strip()
a="  ajay jvnekf  "
print(a)
print(a.strip())
print(a.lstrip())
print(a.rstrip())
```

```python
### split()
a="ajay and 1238@"
print(a.split())
print(a.split("ajay"))
print(a.split('@'))
```

```python
##replace()
a="ajay ansd ajay"
print(a.replace("ajay","jay"))
```

```python
##join() iterable
a=['admwf','a','jay']
print(" ".join(a))
```
